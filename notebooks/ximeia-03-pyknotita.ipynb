### Β5. Παιχνίδι: μάντεψε!

Ο υπολογιστής διαλέγει στην τύχη ένα υλικό και ένα υγρό. Εσύ λες αν θα επιπλεύσει. Η συνάρτηση `ti_kanei` της Β2 ελέγχει. Πέντε γύροι — πόσα βρίσκεις;

Η συνάρτηση `random.choice` διαλέγει έναν αριθμό θέσης στην τύχη, ώστε υλικό και πυκνότητα να μένουν μαζί.

```python
ola_ylika = ["Φελλός", "Πάγος", "Ξύλο", "Αλουμίνιο", "Σίδηρος", "Χρυσός", "Κερί"]
r_ola     = [0.24, 0.92, 0.50, 2.70, 7.87, 19.3, 0.90]
ola_ygra  = ["Οινόπνευμα", "Ελαιόλαδο", "Νερό", "Γλυκερίνη", "Μέλι", "Υδράργυρος"]
r_ola_ygra = [0.79, 0.92, 1.00, 1.26, 1.42, 13.6]

skor = 0
for gyros in range(5):
    i = random.choice(range(len(ola_ylika)))
    j = random.choice(range(len(ola_ygra)))
    apantisi = input(f"Γύρος {gyros + 1}: {ola_ylika[i]} σε {ola_ygra[j]}. Επιπλέει; (ν/ο) ")
    sosto = ti_kanei(r_ola[i], r_ola_ygra[j])
    if (apantisi == "ν" and sosto == "επιπλέει") or (apantisi == "ο" and sosto != "επιπλέει"):
        skor += 1
        print("  ✓ Σωστά!", sosto)
    else:
        print("  ✗ Όχι —", sosto, f"({r_ola[i]} έναντι {r_ola_ygra[j]})")

print(f"Σκορ: {skor} / 5")
```
